# Analyze Dataset

This notebook gives an overview of the dataset distribution, aiming to provide statistics. 

```python
from pathlib import Path
import os
from lib.data.dicom import DicomDataset

datasets = [
    "MIDRC-RICORD-1A",
    "MIDRC-RICORD-1C",
    "LIDC-IDRI",
]
DATA_ROOT = Path("../data")

dataset = DicomDataset(root_dir=DATA_ROOT, accepted_datasets=datasets)
dataset.load()
```

```python

ct_count = 0
xrays_count = 0
multiviews_count = 0
xrays_and_multiviews_count = 0
for patient in dataset:
    #print(dataset[patient])
    cts = dataset[patient]["cts"]
    ct_count = ct_count + len(cts)

    xrays = dataset[patient]["xrays"]
    multiviews = dataset[patient]["multiview"]
    xrays_count = xrays_count + len(xrays)
    multiviews_count = multiviews_count + len(multiviews)
    xrays_and_multiviews_count = xrays_and_multiviews_count + len(xrays) + len(multiviews)

print(f"CTs: {ct_count}")
print(f"X-rays: {xrays_count}")
print(f"Multiviews: {multiviews_count}")
print(f"X-rays and Multiviews: {xrays_and_multiviews_count}")
print(f"Total: {ct_count + xrays_and_multiviews_count}")
print(len(dataset.keys()))
```

# Processed Dataset Statistics

```python
from pathlib import Path
import h5py
DATA_ROOT = Path("preprocess/GAN-dataset")
# iterates over all .h5 files and prints their keys
file_count = len(list(DATA_ROOT.glob("*.h5")))
ct_count = 0
frontal_count = 0
lateral_count = 0
for file in DATA_ROOT.glob("*.h5"):
    with h5py.File(file, "r") as f:
        if "ct" in f.keys():
            ct_count += 1
        if "real_xray1" in f.keys():
            frontal_count += 1
        if "real_xray2" in f.keys():
            lateral_count += 1

print(f"Total files: {file_count}")
print(f"CTs: {ct_count}")
print(f"Frontal X-rays: {frontal_count}")
print(f"Lateral X-rays: {lateral_count}")
```
